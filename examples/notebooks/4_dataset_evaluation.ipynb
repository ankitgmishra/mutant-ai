# Dataset Evaluation

Easily load JSON or CSV files into TestCases and evaluate them.

```
import json
from mutant.datasets import load_test_cases
from mutant.eval import evaluate
from mutant.eval.metrics import Correctness
from mutant.providers import OllamaProvider

provider = OllamaProvider(model="llama3.2")

# Let's create a dummy dataset on disk first
dummy_data = [
    {"input": "2+2", "expected_output": "4", "actual_output": "4"},
    {"input": "Who wrote Hamlet?", "expected_output": "Shakespeare", "actual_output": "Hemingway"}
]
with open("my_dataset.json", "w") as f:
    json.dump(dummy_data, f)
```

```
# 1-Line Load!
dataset = load_test_cases("my_dataset.json")
print(f"Loaded {len(dataset)} cases!")
```

```
report = await evaluate(dataset, metrics=[Correctness(provider)])
report.display(html_path="dataset_report.html")
```
