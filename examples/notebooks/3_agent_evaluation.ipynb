# Agent Evaluation

This notebook demonstrates how to evaluate an Agent's tool selections against a Golden Trajectory.

```
from mutant.eval import TestCase, evaluate
from mutant.eval.metrics import ToolSelection, ToolArgumentCorrectness
from mutant.providers import OllamaProvider

provider = OllamaProvider(model="llama3.2")
```

```
agent_dataset = [
    TestCase(
        input="Find flights to NYC",
        expected_tools=[{"name": "search_flights", "arguments": {"destination": "NYC"}}],
        tools_called=[{"name": "search_flights", "arguments": {"destination": "NYC"}}]
    )
]
```

```
report = await evaluate(
    agent_dataset,
    metrics=[ToolSelection(provider), ToolArgumentCorrectness(provider)]
)

report.display(html_path="agent_report.html")
```
