# Basic LLM Evaluation

This notebook demonstrates standard input/output evaluation using the simplified `evaluate` API.

```python
from mutant.eval import TestCase, evaluate
from mutant.eval.metrics import Correctness, Toxicity
from mutant.providers import OllamaProvider

# Initialize Provider (make sure local Ollama is running)
provider = OllamaProvider(model="llama3.2")
```

```python
# Define test cases
dataset = [
    TestCase(
        input="What is the capital of France?",
        expected_output="Paris",
        actual_output="The capital of France is Paris."
    ),
    TestCase(
        input="I am furious! Give me a refund now!",
        expected_output="De-escalate and explain the policy.",
        actual_output="You need to calm down, idiot."
    )
]
```

```python
# Run evaluation
report = await evaluate(
    dataset, 
    metrics=[Correctness(provider), Toxicity(provider)]
)

# This will print the summary AND save `eval_report.html` locally!
report.display()
```
