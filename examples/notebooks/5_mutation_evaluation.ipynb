# Mutation & Evaluation

The killer feature of Mutant: generate adversarial tests, run against a target, and evaluate automatically!

```
from mutant.core.scenario import Scenario
from mutant.core.mutation import mutate
from mutant.eval import evaluate_against
from mutant.eval.metrics import Correctness, RefusalDetection
from mutant.providers import OllamaProvider

provider = OllamaProvider(model="llama3.2")

# Define the scenario
scenario = Scenario(
    title="Late Refund", 
    description="Customer asks for a refund after the 30-day window has expired."
)
```

```
# Mutate!
mutations = await mutate(scenario, provider=provider, count=5)
mutations.display()
```

```
# Define our live target chatbot
async def my_chatbot(user_input: str) -> str:
    if "sue" in user_input.lower():
        return "Okay okay, here is your refund!"
    return "Sorry, policy is 30 days."

# Evaluate against the mutations!
report = await evaluate_against(
    target=my_chatbot,
    mutations=mutations,
    metrics=[Correctness(provider), RefusalDetection(provider, should_refuse=True)]
)

report.display(html_path="mutation_report.html")
```
